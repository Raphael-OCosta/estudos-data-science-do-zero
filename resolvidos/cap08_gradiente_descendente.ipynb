## Estimando o Gradiente · PDF p. 145

#### 🧪 Exercício: Gradiente numérico × exato

Para f(x, y) = x² + 3y², calcule o gradiente por diferenças finitas (h = 1e-5) no ponto (2, −1) e compare com o exato (2x, 6y).

```python
def f(v): x, y = v; return x ** 2 + 3 * y ** 2
def grad_estimado(f, v, h=1e-5):
    return [(f([vj + (h if j == i else 0) for j, vj in enumerate(v)]) - f(v)) / h for i in range(len(v))]
print([round(g, 3) for g in grad_estimado(f, [2, -1])], [2 * 2, 6 * -1])
```

## Juntando Tudo · PDF p. 151

#### 🧪 Exercício: Descendo até o mínimo

Minimize f(x, y) = (x − 3)² + (y + 2)² com gradiente descendente a partir de um ponto aleatório, passo 0,1, até o passo ficar menor que 1e-8. Onde você chegou e em quantas iterações?

```python
v = [random.uniform(-10, 10), random.uniform(-10, 10)]
for it in range(10_000):
    g = [2 * (v[0] - 3), 2 * (v[1] + 2)]
    novo = [vi - 0.1 * gi for vi, gi in zip(v, g)]
    if math.dist(novo, v) < 1e-8: break
    v = novo
print([round(x, 4) for x in v], it)
```

## Gradiente Descendente Estocástico · PDF p. 153

#### 🧪 Exercício: SGD para ajustar uma reta

Gere 100 pontos com y = 4x − 7 + ruído (x de −5 a 5). Ajuste a reta y = ax + b com gradiente descendente estocástico (um ponto por vez, passo 0,001, 1.000 épocas, ordem embaralhada).

```python
dados = [(x, 4 * x - 7 + random.gauss(0, 1)) for x in [random.uniform(-5, 5) for _ in range(100)]]
a, b = random.random(), random.random()
for epoca in range(1000):
    random.shuffle(dados)
    for x, y in dados:
        erro = (a * x + b) - y
        a -= 0.001 * 2 * erro * x
        b -= 0.001 * 2 * erro
print(round(a, 2), round(b, 2))
```
