#### 🧪 Exercício: Quem tem mais amigos?

Com a lista de amizades `[(0, 1), (0, 2), (1, 2), (1, 3), (2, 3), (3, 4), (4, 5), (5, 6), (5, 7), (6, 8), (7, 8), (8, 9)]`, monte um dicionário `amigos[id] = [ids]`, calcule o número médio de amigos e mostre os 3 usuários mais conectados.

```python
amizades = [(0, 1), (0, 2), (1, 2), (1, 3), (2, 3), (3, 4), (4, 5), (5, 6), (5, 7), (6, 8), (7, 8), (8, 9)]
amigos = defaultdict(list)
for a, b in amizades:
    amigos[a].append(b); amigos[b].append(a)
print("média:", sum(len(v) for v in amigos.values()) / len(amigos))
print(sorted(amigos, key=lambda u: len(amigos[u]), reverse=True)[:3])
```

#### 🧪 Exercício: Amigos de amigos

Escreva `sugestoes(u)`: os amigos dos amigos de `u` que ainda não são amigos dele (nem ele mesmo), com quantos amigos em comum cada um tem (`Counter`). Teste com o usuário 3.

```python
amizades = [(0, 1), (0, 2), (1, 2), (1, 3), (2, 3), (3, 4), (4, 5), (5, 6), (5, 7), (6, 8), (7, 8), (8, 9)]
amigos = defaultdict(set)
for a, b in amizades:
    amigos[a].add(b); amigos[b].add(a)
def sugestoes(u):
    return Counter(ff for f in amigos[u] for ff in amigos[f] if ff != u and ff not in amigos[u])
print(sugestoes(3))
```
