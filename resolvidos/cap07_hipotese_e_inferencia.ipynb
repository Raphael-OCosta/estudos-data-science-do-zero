## Exemplo: Lançar Uma Moeda · PDF p. 127

#### Exercício: A moeda é honesta?

Uma moeda deu 560 caras em 1.000 lançamentos. Use a aproximação normal (μ = np, σ = √(np(1−p))) para calcular o p-valor bicaudal de H₀: p = 0,5.

```python
def normal_cdf(x, mu=0, sigma=1): return (1 + math.erf((x - mu) / (math.sqrt(2) * sigma))) / 2
n, caras, p = 1000, 560, 0.5
mu, sigma = n * p, math.sqrt(n * p * (1 - p))
p_valor = 2 * (1 - normal_cdf(caras - 0.5, mu, sigma))
print(round(p_valor, 5))  # bem menor que 0,05: rejeitamos H0
```

## P-Hacking · PDF p. 134

#### Exercício: P-hacking por simulação

Faça 1.000 "experimentos" com uma moeda honesta (1.000 lançamentos cada) e conte em quantos o p-valor ficou abaixo de 5%. O que isso mostra sobre testar muitas hipóteses?

```python
def normal_cdf(x, mu=0, sigma=1): return (1 + math.erf((x - mu) / (math.sqrt(2) * sigma))) / 2
def p_valor(caras, n=1000):
    mu, s = n / 2, math.sqrt(n) / 2
    return 2 * (1 - normal_cdf(abs(caras - mu), 0, s))
falsos = sum(p_valor(sum(random.random() < 0.5 for _ in range(1000))) < 0.05 for _ in range(1000))
print(falsos)  # ~50: 5% de "descobertas" mesmo sem efeito nenhum
```

## Exemplo: Executando um Teste A/B · PDF p. 135

#### Exercício: Teste A/B de anúncios

O anúncio A teve 200 cliques em 1.000 visualizações; o B teve 150 em 1.000. Calcule a estatística z da diferença de proporções e o p-valor.

```python
def normal_cdf(x): return (1 + math.erf(x / math.sqrt(2))) / 2
def params(n, k): p = k / n; return p, math.sqrt(p * (1 - p) / n)
pa, sa = params(1000, 200); pb, sb = params(1000, 150)
z = (pb - pa) / math.sqrt(sa ** 2 + sb ** 2)
print(round(z, 3), round(2 * (1 - normal_cdf(abs(z))), 4))
```

## Inferência Bayesiana · PDF p. 137

#### Exercício: Priori Beta para a moeda

Com priori Beta(α, β) e k caras em n lançamentos, a posteriori é Beta(α + k, β + n − k). Para 10 caras em 20 lançamentos, calcule a média da posteriori com as prioris Beta(1, 1), Beta(20, 20) e Beta(30, 10).

```python
k, n = 10, 20
for a, b in [(1, 1), (20, 20), (30, 10)]:
    a2, b2 = a + k, b + n - k
    print(f"priori Beta({a},{b}) -> posteriori Beta({a2},{b2}), média = {a2 / (a2 + b2):.3f}")
```
