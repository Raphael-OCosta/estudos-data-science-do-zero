#### 🧪 Exercício: Logística por gradiente

Gere 300 pontos com P(y = 1) = logística(−2 + 1,5x). Ajuste os coeficientes maximizando a log-verossimilhança por subida de gradiente e compare com os valores verdadeiros.

```python
def logistica(t): return 1 / (1 + math.exp(-t))
xs = [random.uniform(-3, 5) for _ in range(300)]
ys = [1 if random.random() < logistica(-2 + 1.5 * x) else 0 for x in xs]
a = b = 0.0
for _ in range(3000):
    ga = sum(y - logistica(a + b * x) for x, y in zip(xs, ys)) / len(xs)
    gb = sum((y - logistica(a + b * x)) * x for x, y in zip(xs, ys)) / len(xs)
    a += 0.1 * ga; b += 0.1 * gb
print(round(a, 2), round(b, 2))
```
