#### 🧪 Exercício: Compreensões e funções

Com uma *list comprehension*, gere os quadrados dos pares de 0 a 20. Com uma *dict comprehension*, mapeie cada palavra de `"o rato roeu a roupa do rei de roma"` para o seu tamanho. Escreva `dobra(f)` que devolve uma função que aplica `f` duas vezes.

```python
print([x * x for x in range(21) if x % 2 == 0])
print({p: len(p) for p in "o rato roeu a roupa do rei de roma".split()})
def dobra(f):
    return lambda x: f(f(x))
print(dobra(lambda x: x + 3)(10))
```

#### 🧪 Exercício: Geradores, zip e *args

Escreva um gerador `fib()` infinito e pegue os 10 primeiros valores com `itertools.islice`. Use `zip` para "descompactar" `[("a", 1), ("b", 2), ("c", 3)]`. Escreva `soma_tudo(*args, **kwargs)` que soma todos os números recebidos.

```python
from itertools import islice
def fib():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b
print(list(islice(fib(), 10)))
letras, numeros = zip(*[("a", 1), ("b", 2), ("c", 3)])
print(letras, numeros)
def soma_tudo(*args, **kwargs):
    return sum(args) + sum(kwargs.values())
print(soma_tudo(1, 2, 3, x=10, y=20))
```
