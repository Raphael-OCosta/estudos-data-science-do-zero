## Recomendando O Que é Popular · PDF p. 379

#### 🧪 Exercício: Os mais populares que você não tem

Com interesses de 6 usuários (listas de tópicos inventadas), conte a popularidade de cada tópico e recomende ao usuário 0 os 3 tópicos mais populares que ele ainda não tem.

```python
interesses = [["python", "estatística"], ["python", "sql", "ml"], ["ml", "estatística", "r"],
              ["sql", "python", "ml"], ["r", "estatística"], ["ml", "deep learning", "python"]]
pop = Counter(t for u in interesses for t in u)
print([t for t, _ in pop.most_common() if t not in interesses[0]][:3])
```

## Filtragem Colaborativa Baseada no Usuário · PDF p. 381

#### 🧪 Exercício: Usuários parecidos (cosseno)

Transforme os interesses em vetores 0/1, calcule a similaridade de cosseno entre o usuário 0 e os outros e recomende os tópicos dos usuários parecidos, pesados pela similaridade.

```python
interesses = [["python", "estatística"], ["python", "sql", "ml"], ["ml", "estatística", "r"],
              ["sql", "python", "ml"], ["r", "estatística"], ["ml", "deep learning", "python"]]
topicos = sorted({t for u in interesses for t in u})
vet = [[1 if t in u else 0 for t in topicos] for u in interesses]
def cos(a, b): return sum(x * y for x, y in zip(a, b)) / math.sqrt(sum(a) * sum(b))
score = defaultdict(float)
for j in range(1, len(vet)):
    s = cos(vet[0], vet[j])
    for t in interesses[j]:
        if t not in interesses[0]: score[t] += s
print(sorted(score.items(), key=lambda kv: -kv[1]))
```
